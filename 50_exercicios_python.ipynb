21. Implemente a busca binária para encontrar um elemento em uma lista ordenada.

```python
def procurar_numero(lista, valor): #cria uma função para procurar um valor na lista
    esquerda = 0 #começa a busca pela primeira posição
    direita = len(lista) - 1 #define a última posição da lista

    while esquerda <= direita: #continua enquanto ainda existir uma parte da lista para procurar
        meio = (esquerda + direita) // 2 #descobre a posição que fica no meio

        if lista[meio] == valor: #verifica se o valor do meio é o que estamos procurando
            return meio #retorna a posição quando encontra

        elif valor > lista[meio]: #verifica se o valor procurado está na parte maior da lista
            esquerda = meio + 1 #descarta a metade da esquerda

        else:
            direita = meio - 1 #descarta a metade da direita

    return -1 #retorna -1 se o número não estiver na lista


numeros = [2, 5, 8, 10, 14, 20, 25] #cria uma lista já organizada

procurado = int(input("Digite um número para procurar: ")) #guarda o número digitado

posicao = procurar_numero(numeros, procurado) #faz a busca e guarda o resultado

if posicao == -1: #verifica se o número não foi encontrado
    print("Número não encontrado.")
else:
    print("Número encontrado na posição:", posicao) #mostra onde ele foi encontrado
```

Explicação: Usei uma função para deixar a busca separada do restante do código. A busca binária funciona olhando primeiro o valor do meio da lista. Se não for o número procurado, ela elimina metade da lista e continua procurando só na parte onde o número pode estar. Fiz dessa forma porque é mais rápido do que verificar todos os números um por um. A lista precisa estar em ordem para esse método funcionar.

22. Crie uma função para criptografar uma mensagem usando a cifra de César.

```python
def codificar(texto, passo): #cria uma função para mudar as letras da mensagem
    letras = "abcdefghijklmnopqrstuvwxyz" #guarda as letras que vão ser usadas
    nova_mensagem = "" #começa uma mensagem vazia para montar o resultado

    for caractere in texto: #passa por cada caractere digitado
        if caractere in letras: #verifica se o caractere é uma letra
            indice = letras.index(caractere) #descobre a posição da letra
            novo_indice = (indice + passo) % 26 #avança a quantidade escolhida no alfabeto
            nova_mensagem = nova_mensagem + letras[novo_indice] #adiciona a nova letra na mensagem

        else:
            nova_mensagem = nova_mensagem + caractere #mantém espaços e outros caracteres

    return nova_mensagem #devolve a mensagem depois da alteração


texto = input("Digite uma mensagem: ") #guarda a mensagem digitada
passo = int(input("Digite o valor do deslocamento: ")) #guarda quantas posições as letras vão andar

resultado = codificar(texto, passo) #chama a função e guarda a mensagem pronta

print("Mensagem criptografada:", resultado) #mostra o resultado
```

Explicação: Criei uma função para fazer a criptografia e coloquei o alfabeto dentro de uma variável. Depois passei por cada letra da mensagem e descobri a posição dela no alfabeto. O valor do deslocamento é somado nessa posição para encontrar a nova letra. Usei % 26 para fazer a contagem voltar ao começo quando passar da última letra do alfabeto. Os espaços e outros caracteres são mantidos normalmente.

23. Crie uma função para descriptografar uma mensagem cifrada com a cifra de César

```python
def decodificar(texto, passo): #cria uma função para recuperar a mensagem original
    letras = "abcdefghijklmnopqrstuvwxyz" #guarda as letras do alfabeto
    mensagem_original = "" #começa uma mensagem vazia para montar o resultado

    for caractere in texto: #passa por cada caractere da mensagem
        if caractere in letras: #verifica se o caractere está no alfabeto
            indice = letras.index(caractere) #descobre a posição atual da letra
            indice_original = (indice - passo) % 26 #volta a quantidade de posições informada
            mensagem_original += letras[indice_original] #adiciona a letra correta no resultado
        else:
            mensagem_original += caractere #mantém espaços e outros caracteres

    return mensagem_original #devolve a mensagem já descriptografada


texto = input("Digite a mensagem criptografada: ") #guarda a mensagem digitada
passo = int(input("Digite o valor do deslocamento: ")) #guarda quantas posições devem ser voltadas

resultado = decodificar(texto, passo) #chama a função e guarda a mensagem original

print("Mensagem descriptografada:", resultado) #mostra o resultado final
```

Explicação: Criei uma função para fazer o processo contrário da criptografia. Em vez de somar o deslocamento, eu subtraio esse valor da posição da letra para voltar ao caractere original. Usei % 26 para continuar funcionando mesmo quando a contagem passa do começo do alfabeto. Os espaços e outros caracteres continuam iguais.

24. Implemente uma função para validar um CPF brasileiro.

```python
def verificar_cpf(cpf_digitado):  # cria a função responsável pela validação

    somente_numeros = cpf_digitado.replace(".", "").replace("-", "")  # remove a formatação

    if len(somente_numeros) != 11:  # verifica a quantidade de dígitos
        return False

    elif somente_numeros == somente_numeros[0] * 11:  # impede CPFs com todos os números iguais
        return False

    calculo1 = 0

    peso = 10  # começa o peso do primeiro cálculo

    for posicao in range(9):
        calculo1 += int(somente_numeros[posicao]) * peso
        peso -= 1  # diminui o peso a cada repetição

    digito1 = (calculo1 * 10) % 11

    if digito1 >= 10:
        digito1 = 0

    if digito1 != int(somente_numeros[9]):
        return False

    calculo2 = 0
    peso = 11  # reinicia o peso para o segundo cálculo

    for posicao in range(10):
        calculo2 += int(somente_numeros[posicao]) * peso
        peso -= 1

    digito2 = (calculo2 * 10) % 11

    if digito2 >= 10:
        digito2 = 0

    if digito2 != int(somente_numeros[10]):
        return False

    return True


cpf_usuario = input("Informe um CPF: ")

if verificar_cpf(cpf_usuario):
    print("CPF válido")
else:
    print("CPF inválido")
```

Explicação: O programa cria uma função para verificar se um CPF é válido. Primeiro, ele remove pontos e traços para trabalhar apenas com os números. Depois, verifica se existem 11 dígitos e se todos eles não são iguais.

Para calcular os dígitos verificadores, foram utilizadas as variáveis peso, calculo1 e calculo2. O peso começa em 10 no primeiro cálculo e vai diminuindo a cada repetição. No segundo cálculo, ele começa em 11. Depois, os dígitos encontrados são comparados com os dois últimos números do CPF.

Esse método foi escolhido para deixar o cálculo mais fácil de acompanhar, pois o valor do peso é reduzido diretamente dentro do for. Também foram usados nomes de variáveis diferentes para deixar mais claro o papel de cada parte do código e evitar repetir a mesma estrutura do código anterior.

25. Classifique uma lista de números em três categorias: positivos, negativos e zeros.

```python
valores = list(map(int, input("Digite alguns números separados por espaço: ").split())) #pega os numeros digitados e coloca em uma lista

positivos = [] #guarda os numeros maiores que zero
negativos = [] #guarda os numeros menores que zero
zeros = [] #guarda os valores iguais a zero

for valor in valores: #passa por cada numero da lista
    if valor > 0: #verifica se o numero é positivo
        positivos.append(valor) #coloca o numero na lista de positivos

    elif valor < 0: #verifica se o numero é negativo
        negativos.append(valor) #coloca o numero na lista de negativos

    else: #se nao for positivo nem negativo, é zero
        zeros.append(valor) #adiciona o zero na lista

print("Positivos:", positivos)
print("Negativos:", negativos)
print("Zeros:", zeros)
```

Explicação: Primeiro pedi vários números de uma vez e usei .split() para separar o que foi digitado. O map(int, ...) transforma cada valor em número inteiro. Depois criei três listas diferentes e usei um for para verificar cada número. Dependendo se ele é maior, menor ou igual a zero, ele é colocado na lista correspondente.

29. Implemente o clássico problema FizzBuzz para número de 1 a 100.

```python
def validar_cpf(cpf):  # cria uma função para validar o CPF

    cpf_limpo = cpf.replace(".", "").replace("-", "")  # remove pontos e traço do CPF

    if len(cpf_limpo) != 11:  # verifica se o CPF possui 11 números
        return False

    elif cpf_limpo == cpf_limpo[0] * 11:  # verifica se todos os números são iguais
        return False

    soma_primeiro = 0  # armazena a soma do primeiro dígito

    for i in range(9):  # percorre os 9 primeiros números
        soma_primeiro += int(cpf_limpo[i]) * (10 - i)

    primeiro_digito = (soma_primeiro * 10) % 11  # calcula o primeiro dígito

    if primeiro_digito == 10:  # verifica se o resultado foi 10
        primeiro_digito = 0

    if primeiro_digito != int(cpf_limpo[9]):  # compara com o dígito do CPF
        return False

    soma_segundo = 0  # armazena a soma do segundo dígito

    for i in range(10):  # percorre os 10 primeiros números
        soma_segundo += int(cpf_limpo[i]) * (11 - i)

    segundo_digito = (soma_segundo * 10) % 11  # calcula o segundo dígito

    if segundo_digito == 10:  # verifica se o resultado foi 10
        segundo_digito = 0

    if segundo_digito != int(cpf_limpo[10]):  # compara com o dígito do CPF
        return False

    return True


cpf = input("Digite um CPF: ")

if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

Explicação: O programa percorre os números de 1 até 100 e verifica cada um usando o resto da divisão. Se o número for divisível por 3 e por 5 ao mesmo tempo, mostra FizzBuzz. Se for apenas por 3, mostra Fizz, e se for apenas por 5, mostra Buzz.

Essa ordem foi escolhida porque a condição de ser divisível pelos dois precisa vir primeiro. Se ela viesse depois, números como 15 poderiam ser identificados apenas como Fizz ou Buzz. O for também facilita porque faz essa verificação automaticamente para todos os números do intervalo.

30. Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra

```python
import random #importa a biblioteca usada para escolher uma palavra aleatoria

lista_palavras = ["banana", "macaco", "computador", "python", "escola"] #guarda as palavras que podem aparecer no jogo

palavra_secreta = random.choice(lista_palavras) #sorteia uma das palavras da lista

acertos = [] #guarda as letras que o jogador acertou
vidas = 6 #define quantas vezes o jogador pode errar

while vidas > 0: #mantem o jogo funcionando enquanto ainda tiver tentativas

    palavra_atual = "" #monta a palavra que vai aparecer para o jogador

    for letra in palavra_secreta: #passa por cada letra da palavra sorteada
        if letra in acertos: #verifica se essa letra ja foi descoberta
            palavra_atual += letra #mostra a letra acertada
        else:
            palavra_atual += "_" #esconde as letras que ainda nao foram acertadas

    print("\nPalavra:", palavra_atual)
    print("Tentativas restantes:", vidas)

    if palavra_atual == palavra_secreta: #confere se a palavra inteira ja foi descoberta
        print("Você ganhou!")
        break #encerra o jogo quando o jogador vence

    letra_digitada = input("Digite uma letra: ").lower() #pega o palpite do jogador

    if letra_digitada in palavra_secreta: #verifica se a letra existe na palavra
        acertos.append(letra_digitada) #guarda a letra que foi acertada
        print("Você acertou!")
    else:
        vidas -= 1 #tira uma tentativa quando a letra estiver errada
        print("Você errou!")

if vidas == 0: #entra aqui se todas as tentativas acabarem
    print("Você perdeu!")
    print("A palavra era:", palavra_secreta)
```

Explicação: Primeiro criei uma lista de palavras e usei random.choice() para escolher uma delas sem o jogador saber qual é. Depois criei uma lista para guardar as letras acertadas e defini 6 tentativas. O while mantém o jogo rodando até a pessoa acertar ou perder todas as tentativas. Dentro dele, usei um for para montar a palavra com letras e traços. Quando a letra digitada existe na palavra, ela é guardada nos acertos. Quando não existe, o jogador perde uma tentativa.

31. Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
valores = list(map(int, input("Digite os números separados por espaço: ").split())) #pega os numeros digitados e coloca em uma lista

for i in range(len(valores)): #repete o processo de organização da lista
    for j in range(len(valores) - 1 - i): #passa pelos valores que ainda precisam ser comparados

        if valores[j] > valores[j + 1]: #verifica se o numero da esquerda é maior que o da direita
            valores[j], valores[j + 1] = valores[j + 1], valores[j] #troca os dois de lugar

print("Lista ordenada:", valores) #mostra a lista do menor para o maior
```

Explicação: Primeiro pedi os números e coloquei todos em uma lista. Depois usei dois for porque esse método precisa comparar os valores várias vezes. Quando um número está maior que o próximo, os dois trocam de posição. Esse processo vai se repetindo até a lista ficar toda em ordem crescente.

32. Crie um jogo da velha para dois jogadores.

```python
tabuleiro = [" "] * 9 #cria 9 espaços vazios para montar o tabuleiro


def mostrar(): #cria uma função só para exibir o jogo
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print("--+---+--")
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print("--+---+--")
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])


def ganhou(): #cria uma função para conferir se alguém venceu
    possibilidades = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ] #guarda todas as formas possíveis de fazer uma sequência de 3

    for posicoes in possibilidades: #passa por cada possibilidade de vitória
        p1, p2, p3 = posicoes

        if tabuleiro[p1] == tabuleiro[p2] == tabuleiro[p3] != " ": #confere se as 3 posições têm o mesmo símbolo
            return True

    return False #retorna falso se ninguém tiver ganhado


jogador = "X" #define quem começa a partida

for jogada in range(9): #o jogo pode ter no máximo 9 jogadas

    mostrar()

    escolha = int(input(f"Jogador {jogador}, escolha uma posição de 1 a 9: ")) - 1 #pega a posição escolhida

    while tabuleiro[escolha] != " ": #não deixa escolher um espaço que já foi usado
        print("Essa posição já está ocupada!")
        escolha = int(input("Escolha outra posição: ")) - 1

    tabuleiro[escolha] = jogador #coloca X ou O na posição escolhida

    if ganhou(): #verifica se a jogada completou uma sequência
        mostrar()
        print(f"Jogador {jogador} venceu!")
        break

    if jogador == "X": #troca para o outro jogador
        jogador = "O"
    else:
        jogador = "X"

else:
    mostrar()
    print("Empate!") #aparece se as 9 posições forem usadas sem vencedor
```

Explicação: Criei o tabuleiro usando uma lista com 9 espaços, porque cada espaço representa uma posição do jogo da velha. Depois fiz uma função para mostrar o tabuleiro e outra para conferir se algum jogador conseguiu formar uma sequência de três símbolos iguais. O jogo repete as jogadas até alguém ganhar ou até as 9 posições serem preenchidas. A cada jogada, o programa também verifica se o lugar escolhido já está ocupado e troca o jogador entre `X` e `O`.

37. Implemente uma função recursiva para calcular o fatorial de um número.

```python
def calcular_fatorial(valor): #cria uma função para calcular o fatorial

    if valor == 0 or valor == 1: #define o ponto em que a função deve parar
        return 1

    return valor * calcular_fatorial(valor - 1) #multiplica o valor atual pelo anterior


numero = int(input("Digite um número: ")) #guarda o número informado pelo usuario

fatorial = calcular_fatorial(numero) #chama a função e guarda o resultado

print("Fatorial:", fatorial) #mostra o resultado final
```

Explicação: Criei uma função para calcular o fatorial usando recursão, ou seja, a própria função vai sendo chamada novamente com um número menor. Quando chega em 0 ou 1, ela retorna 1 e para de se repetir. Depois, os valores vão sendo multiplicados até chegar no resultado final.

46. Calcule a média e o desvio padrão de uma lista de números.

48. Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
valores = list(map(int, input("Digite os números separados por espaço: ").split())) #guarda os numeros digitados em uma lista

for i in range(1, len(valores)): #começa a percorrer a lista a partir do segundo numero
    numero_atual = valores[i] #guarda o numero que está sendo analisado
    posicao = i - 1 #começa comparando com o numero anterior

    while posicao >= 0 and valores[posicao] > numero_atual: #continua enquanto o numero anterior for maior
        valores[posicao + 1] = valores[posicao] #move o numero maior uma posição para frente
        posicao = posicao - 1 #volta para comparar com o numero anterior

    valores[posicao + 1] = numero_atual #coloca o numero na posição certa

print("Lista ordenada:", valores) #mostra a lista organizada
```

Explicação:Primeiro pedi os números e coloquei todos em uma lista. Depois comecei a verificar os valores a partir do segundo número. Cada valor é comparado com os que vêm antes dele e, se estiver fora de ordem, os maiores vão sendo movidos para frente. No final, o número atual é colocado na posição certa. Esse processo se repete até a lista ficar em ordem crescente.

49. Calcule a soma dos dígitos de um número inteiro fornecido.

```python
valor = int(input("Digite um número inteiro: ")) #guarda o numero digitado pelo usuario

total = 0 #começa a soma dos algarismos em zero

for numero in str(abs(valor)): #transforma o numero em texto e passa por cada algarismo
    total = total + int(numero) #transforma o algarismo em numero novamente e soma ao total

print("Soma dos dígitos:", total) #mostra o resultado da soma
```

Explicação:Primeiro pedi um número inteiro para o usuário. Depois usei abs() para evitar problema caso o número seja negativo e transformei ele em texto com str(), assim consegui passar por cada dígito separadamente. Em seguida, transformei cada dígito novamente em número e fui somando todos eles. No final, mostrei o resultado.
