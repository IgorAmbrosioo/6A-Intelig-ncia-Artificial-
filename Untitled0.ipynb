19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o
angulo entre eles (produto escalar normalizado).

```python
import numpy as np  # Importa a biblioteca NumPy
array1 = np.random.randint(1, 101, 20)  # Cria o primeiro array
array2 = np.random.randint(1, 101, 20)  # Cria o segundo array
produto = np.dot(array1, array2)  # Calcula o produto escalar
norma1 = np.sqrt(np.sum (array1 ** 2))  # Calcula o tamanho do primeiro array
norma2 = np.sqrt(np.sum (array2 ** 2))  # Calcula o tamanho do segundo array
angulo = np.degrees(np.arccos(produto / (norma1 * norma2)))  # Calcula o ângulo
print ("Array 1:", array1)
print ("Array 2:", array2)
print ("Ângulo:", angulo)
```

Explicação: Criei dois arrays com 20 números inteiros aleatórios. Depois calculei o produto escalar entre eles, normalizei usando o tamanho dos arrays e por último encontrei o ângulo em graus. O material também apresenta np.dot() para calcular o produto escalar entre vetores.

a 10.

```python
from scipy.integrate import solve_ivp
import numpy as np
def equacao(x, y):  # Define a equação diferencial
    return x - y
x = np.linspace(0, 10, 100)  # Cria os pontos de 0 até 10
solucao = solve_ivp(equacao, [0, 10], [1], t_eval=x)  # Resolve a equação
print ("Valores de x:", solucao.t)
print ("Valores de y:", solucao.y[0])
```

2+ 4x+ 1.

```python
from scipy.optimize import minimize
def f(x):  # Define a função
   return -x**2 + 4*x + 1
resultado = minimize(lambda x: -f(x), 0)  # Multiplica por -1 para encontrar o máximo
x_max = resultado.x[0]
valor_max = f(x_max)
print ("Valor de x:", x_max)
print ("Valor máximo:", valor_max)
```

8. Plote um grafico de linha com intervalos de confianca (sombramento) para
a funcao y = sin(x).


```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 100)
y = np.sin(x)
limite = 0.2  # Define o intervalo
plt.plot(x, y)
plt.fill_between(x, y - limite, y + limite, alpha=0.3)  # Cria o sombreamento

plt.title("y = sin(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

Explicação: Criei os valores de x e calculei sin(x). Depois usei fill_between() para criar uma área sombreada de 0,2 para cima e para baixo da linha, representando o intervalo de confiança.
